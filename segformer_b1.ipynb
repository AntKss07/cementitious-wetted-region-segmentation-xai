# SegFormer-B1 with MiT-B1 Encoder for Cementitious Sorptivity Segmentation

This notebook implements the complete **SegFormer-B1** deep learning pipeline for capillary water absorption segmentation in cementitious materials, replicating the exact experimental protocol from the **U-Net++** and **DeepLabV3+** baselines for a rigorous, fair cross-architecture comparison.

### Multi-Architecture Research Benchmark:
This implementation forms the fourth pillar of our comprehensive benchmark:
1. **FPN + EfficientNet-B2**: Feature Pyramid CNN baseline (Nature Communications, 2024).
2. **U-Net++ + ResNet-34**: Nested dense skip pathway CNN.
3. **DeepLabV3+ + ResNet-50**: Atrous spatial pyramid pooling (ASPP) CNN.
4. **SegFormer-B1 + MiT-B1**: Hierarchical Mix Transformer with lightweight All-MLP decoder.

---

### Key Architectural Characteristics of SegFormer-B1:
1. **Hierarchical Mix Transformer (MiT-B1) Encoder**:
   - Unlike standard Vision Transformers (ViT) that process fixed non-overlapping patches at a single scale, MiT-B1 uses **overlapping patch embedding** to generate multi-scale feature representations across 4 hierarchical stages:
     - Stage 1: $112 \times 112 \times 64$
     - Stage 2: $56 \times 56 \times 128$
     - Stage 3: $28 \times 28 \times 320$
     - Stage 4: $14 \times 14 \times 512$
2. **Positional-Encoding-Free Design**:
   - Uses $3 \times 3$ depthwise convolutions in the feed-forward networks (Mix-FFN) to dynamically interpolate positional information, allowing arbitrary input resolutions without interpolation artifacts.
3. **Efficient Self-Attention**:
   - Reduces the sequence length of keys and values using spatial reduction ratios ($R$), decreasing computational complexity from $O(N^2)$ to $O(N^2 / R)$.
4. **Lightweight All-MLP Decoder**:
   - Replaces computationally heavy decoder networks with simple linear projections that align channel dimensions, upsample feature maps to $1/4$ resolution ($112 \times 112$), concatenate, fuse via an MLP, and predict 1-channel logits.

---

### Strict Experimental Controls for Fair Comparison:
- **Same Locked Dataset & Partitions**: Exactly preserves `video_split_assignment.csv` (19 Train, 6 Val, 3 Test videos).
- **Same Consensus Masks**: Uses the exact consensus ground truth ($\ge 2$ of 4 annotators agreeing on foreground).
- **Same Resolution**: $448 \times 448 \times 3$ RGB with ImageNet normalization.
- **Same Augmentation**: Identical synchronized spatial (H-flip, V-flip, random crop) and photometric (color jitter) transforms.
- **Same Loss Function**: Combined Binary Cross-Entropy with Logits and Dice Loss (`BCEDiceLoss`, 0.5 BCE + 0.5 Dice).
- **Same Evaluation Protocol**: Evaluates untouched test videos (`video 24`, `video 1`, `video 21`), reporting Dice, IoU, Precision, Recall, and downstream physical WAR metrics (MAE, RMSE, and $R^2$ via `sklearn.metrics.r2_score`).
- **Independent Checkpoints**: Saved in `evaluation/segformer_b1/` and `checkpoints/segformer_b1/` to prevent overwriting prior models.

## Section 3: Consensus Mask Verification
Verifies that the consensus ground-truth masks cached in `video-frame/video-frame/consensus_masks/` are intact and identical to those used by U-Net++ and DeepLabV3+.
- Formulated via majority voting ($\ge 2$ of 4 annotators agreeing on foreground wet pixels).
- Eliminates individual annotator subjective border noise while providing a single canonical ground truth for downstream WAR comparison.

```python
def verify_consensus_masks(frames_dir: Path, consensus_dir: Path) -> None:
    """Verifies that all 1,250 frames across 28 videos have valid cached consensus masks."""
    video_folders = sorted(
        [d.name for d in frames_dir.iterdir() if d.is_dir()],
        key=lambda x: int(x.split()[-1]) if x.split()[-1].isdigit() else x
    )
    total_frames = 0
    total_masks = 0
    for vid in video_folders:
        frames = list((frames_dir / vid).glob("*.jpg"))
        masks = list((consensus_dir / vid).glob("*.png"))
        total_frames += len(frames)
        total_masks += len(masks)
        assert len(frames) == len(masks), f"Mismatch in {vid}: {len(frames)} frames vs {len(masks)} masks"

    print(f"Consensus Mask Verification Passed:")
    print(f"  Total video folders: {len(video_folders)}")
    print(f"  Total frames verified: {total_frames}")
    print(f"  Total consensus masks verified: {total_masks}")

verify_consensus_masks(FRAMES_DIR, CONSENSUS_DIR)
```

## Section 4: Dataset Class & DataLoaders
Implements `SorptivityVideoDataset` and constructs DataLoaders for train, val, and test splits.
- **Identical Augmentations**: Random horizontal flip, vertical flip, random crop/scale, and photometric color jitter (brightness/contrast).
- **Standardized Preprocessing**: $448 \times 448$ bilinear resize for images, nearest-neighbor resize for masks, and ImageNet normalization.

```python
class SorptivityVideoDataset(Dataset):
    """PyTorch Dataset for paired sorptivity video frames and segmentation masks."""
    def __init__(
        self,
        split: str,
        split_df: pd.DataFrame,
        frames_dir: Path = FRAMES_DIR,
        consensus_dir: Path = CONSENSUS_DIR,
        image_size: int = IMAGE_SIZE,
        is_train: bool = False,
    ):
        self.split = split
        self.image_size = image_size
        self.is_train = is_train

        # Filter videos belonging strictly to this split partition
        videos_in_split = split_df[split_df["split"] == split]["video"].tolist()
        self.samples: List[Tuple[Path, Path, str, str]] = []

        for vid in videos_in_split:
            vid_frames = sorted((frames_dir / vid).glob("*.jpg"))
            for f_path in vid_frames:
                mask_path = consensus_dir / vid / f"{f_path.stem}.png"
                if mask_path.exists():
                    self.samples.append((f_path, mask_path, vid, f_path.stem))

        if len(self.samples) == 0:
            raise RuntimeError(f"No samples found for split: '{split}'")

    def __len__(self) -> int:
        return len(self.samples)

    def __getitem__(self, idx: int) -> Tuple[torch.Tensor, torch.Tensor, str, str]:
        img_path, mask_path, video_name, stem = self.samples[idx]

        image = Image.open(img_path).convert("RGB")
        mask = Image.open(mask_path).convert("L")

        # Resize image (bilinear) and mask (nearest neighbor)
        image = TF.resize(image, [self.image_size, self.image_size])
        mask = TF.resize(mask, [self.image_size, self.image_size],
                         interpolation=TF.InterpolationMode.NEAREST)

        # Synchronized Joint Augmentations for Training Split
        if self.is_train:
            # 1. Random Horizontal Flip (physically plausible)
            if random.random() < 0.5:
                image = TF.hflip(image)
                mask = TF.hflip(mask)

            # 3. Random Resized Crop / Scale variation
            pad = 24
            image = TF.pad(image, pad)
            mask = TF.pad(mask, pad)
            i, j, h, w = transforms.RandomResizedCrop.get_params(
                image, scale=(0.85, 1.0), ratio=(1.0, 1.0)
            )
            image = TF.resized_crop(image, i, j, h, w, [self.image_size, self.image_size])
            mask = TF.resized_crop(mask, i, j, h, w, [self.image_size, self.image_size],
                                   interpolation=TF.InterpolationMode.NEAREST)

            # 4. Photometric Jitter (brightness & contrast to combat specular reflection noise)
            brightness_factor = random.uniform(0.8, 1.2)
            contrast_factor = random.uniform(0.8, 1.2)
            image = TF.adjust_brightness(image, brightness_factor)
            image = TF.adjust_contrast(image, contrast_factor)

        # Convert to Tensors
        img_tensor = TF.to_tensor(image)  # Shape: (3, H, W), float in [0.0, 1.0]
        img_tensor = TF.normalize(img_tensor, mean=IMAGENET_MEAN, std=IMAGENET_STD)

        mask_np = np.array(mask)
        mask_tensor = torch.from_numpy((mask_np > 127).astype(np.float32)).unsqueeze(0)  # Shape: (1, H, W)

        return img_tensor, mask_tensor, video_name, stem


def get_dataloaders(
    split_df: pd.DataFrame,
    batch_size: int = BATCH_SIZE,
    num_workers: int = NUM_WORKERS
) -> Tuple[DataLoader, DataLoader, DataLoader]:
    """Constructs DataLoaders for train, val, and test partitions."""
    train_ds = SorptivityVideoDataset("train", split_df, is_train=True)
    val_ds = SorptivityVideoDataset("val", split_df, is_train=False)
    test_ds = SorptivityVideoDataset("test", split_df, is_train=False)

    train_loader = DataLoader(
        train_ds, batch_size=batch_size, shuffle=True,
        num_workers=num_workers, pin_memory=torch.cuda.is_available(), drop_last=True
    )
    val_loader = DataLoader(
        val_ds, batch_size=batch_size, shuffle=False,
        num_workers=num_workers, pin_memory=torch.cuda.is_available(), drop_last=False
    )
    test_loader = DataLoader(
        test_ds, batch_size=batch_size, shuffle=False,
        num_workers=num_workers, pin_memory=torch.cuda.is_available(), drop_last=False
    )

    print(f"DataLoaders initialized:")
    print(f"  Train: {len(train_ds)} frames ({len(train_loader)} batches)")
    print(f"  Val:   {len(val_ds)} frames ({len(val_loader)} batches)")
    print(f"  Test:  {len(test_ds)} frames ({len(test_loader)} batches)")
    return train_loader, val_loader, test_loader

train_loader, val_loader, test_loader = get_dataloaders(split_df, batch_size=BATCH_SIZE)
```

## Section 5: SegFormer-B1 Architecture with Pretrained MiT-B1 Encoder
Instantiates the **SegFormer-B1** architecture using `smp.Segformer`:
- **Encoder**: Hierarchical Mix Transformer (MiT-B1) pretrained on ImageNet.
  - Multi-scale feature stages: $112 \times 112$ (Stage 1), $56 \times 56$ (Stage 2), $28 \times 28$ (Stage 3), $14 \times 14$ (Stage 4).
  - Overlapping patch embedding and depthwise convolution-based Mix-FFN.
- **Decoder**: Lightweight All-MLP decoder fusing the multi-scale stages to full $448 \times 448$ resolution.
- **Output**: 1 output logit channel (no activation during training; sigmoid applied during inference).

```python
def build_segformer_model(
    encoder_name: str = "mit_b1",
    encoder_weights: str = "imagenet",
    in_channels: int = 3,
    classes: int = 1
) -> nn.Module:
    """Builds and returns a SegFormer-B1 semantic segmentation model.
    
    Args:
        encoder_name: Backbone architecture (MiT-B1).
        encoder_weights: Pretrained initialization ('imagenet').
        in_channels: 3 for RGB inputs.
        classes: 1 for binary segmentation (wet vs dry).
        
    Returns:
        Configured PyTorch nn.Module.
    """
    model = smp.Segformer(
        encoder_name=encoder_name,
        encoder_weights=encoder_weights,
        in_channels=in_channels,
        classes=classes,
    )
    return model

model = build_segformer_model().to(DEVICE)
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"SegFormer-B1 (MiT-B1) initialized successfully.")
print(f"Trainable parameters: {num_params:,}")
```

## Section 6: Hierarchical Feature Stages & Tensor Shape Verification
Inspects the MiT-B1 multi-scale feature hierarchy and verifies tensor shapes through the complete network:
- Input: `[B, 3, 448, 448]`
- Encoder Stage 1: `[B, 64, 112, 112]`
- Encoder Stage 2: `[B, 128, 56, 56]`
- Encoder Stage 3: `[B, 320, 28, 28]`
- Encoder Stage 4: `[B, 512, 14, 14]`
- Final Logits: `[B, 1, 448, 448]`

```python
def verify_segformer_tensor_shapes(model: nn.Module) -> None:
    """Verifies the multi-scale tensor shapes produced by MiT-B1 and the All-MLP decoder."""
    model.eval()
    dummy_input = torch.randn(2, 3, IMAGE_SIZE, IMAGE_SIZE, device=DEVICE)
    with torch.no_grad():
        features = model.encoder(dummy_input)
        logits = model(dummy_input)

    print("MiT-B1 Multi-Scale Feature Hierarchy:")
    # In SMP, features[0] is input, features[1] is dummy stem, features[2..5] are the 4 MiT stages
    mit_stage_names = ["Stage 1 (~112x112x64)", "Stage 2 (~56x56x128)", "Stage 3 (~28x28x320)", "Stage 4 (~14x14x512)"]
    for i, stage_name in enumerate(mit_stage_names):
        feat = features[i + 2]
        print(f"  {stage_name} -> Tensor Shape: {list(feat.shape)}")

    print(f"\nFinal Decoder Output:")
    print(f"  Input Shape  : {list(dummy_input.shape)}")
    print(f"  Logits Shape : {list(logits.shape)} (1 channel per pixel)")
    assert logits.shape == (2, 1, IMAGE_SIZE, IMAGE_SIZE), "Mismatch in SegFormer output shape!"
    print("Tensor shape verification: PASSED!")

verify_segformer_tensor_shapes(model)
```

## Section 7: Loss Function, Evaluation Metrics & Physical Domain Constraints
Replicates the exact loss and evaluation definitions from the U-Net++ and DeepLabV3+ benchmarks:
1. **Hybrid BCE + Dice Loss**: `BCEDiceLoss(bce_weight=0.5, dice_weight=0.5)`.
2. **Segmentation Metrics**: Batch-level IoU, Dice, Precision, Recall, and Pixel Accuracy.
3. **Physical Waterline Post-Processing**: `enforce_waterline_physics()` enforces bottom-edge reservoir contiguity and column waterline monotonicity.
4. **Wetted Area Ratio (WAR)**: Calculates $WAR = |M_{wet} \cap M_{ROI}| / |M_{ROI}|$, and computes WAR MAE, RMSE, and $R^2$ using genuine `sklearn.metrics.r2_score`.

```python
class BCEDiceLoss(nn.Module):
    """Hybrid loss combining Binary Cross-Entropy with Logits and Dice Loss."""
    def __init__(self, bce_weight: float = 0.5, dice_weight: float = 0.5, smooth: float = 1.0):
        super().__init__()
        self.bce = nn.BCEWithLogitsLoss()
        self.bce_weight = bce_weight
        self.dice_weight = dice_weight
        self.smooth = smooth

    def forward(self, logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
        bce_loss = self.bce(logits, targets)
        probs = torch.sigmoid(logits)
        intersection = (probs * targets).sum(dim=(1, 2, 3))
        cardinality = probs.sum(dim=(1, 2, 3)) + targets.sum(dim=(1, 2, 3))
        dice_loss = 1.0 - (2.0 * intersection + self.smooth) / (cardinality + self.smooth)
        return self.bce_weight * bce_loss + self.dice_weight * dice_loss.mean()


def compute_batch_metrics(
    preds_binary: torch.Tensor,
    targets_binary: torch.Tensor,
    eps: float = 1e-7
) -> Dict[str, float]:
    """Computes IoU, Dice, Precision, Recall, and Accuracy over a batch of binary masks."""
    tp = (preds_binary * targets_binary).sum(dim=(1, 2, 3))
    fp = (preds_binary * (1.0 - targets_binary)).sum(dim=(1, 2, 3))
    fn = ((1.0 - preds_binary) * targets_binary).sum(dim=(1, 2, 3))
    tn = ((1.0 - preds_binary) * (1.0 - targets_binary)).sum(dim=(1, 2, 3))

    iou = (tp + eps) / (tp + fp + fn + eps)
    dice = (2.0 * tp + eps) / (2.0 * tp + fp + fn + eps)
    precision = (tp + eps) / (tp + fp + eps)
    recall = (tp + eps) / (tp + fn + eps)
    accuracy = (tp + tn) / (tp + fp + fn + tn + eps)

    return {
        "iou": float(iou.mean().item()),
        "dice": float(dice.mean().item()),
        "precision": float(precision.mean().item()),
        "recall": float(recall.mean().item()),
        "accuracy": float(accuracy.mean().item()),
    }


def enforce_waterline_physics(
    binary_mask: np.ndarray,
    bottom_margin: int = BOTTOM_MARGIN,
    close_kernel_size: Tuple[int, int] = (5, 7),
) -> np.ndarray:
    """Enforces physical sorptivity boundary constraints on a binary mask.
    
    1. Bottom Contiguity: Water enters exclusively from the bottom reservoir.
       Eliminates floating reflection blobs and surface artifacts.
    2. Column Monotonicity: Within each column, only pixels contiguous to the bottom waterline are valid.
    """
    mask = (binary_mask > 0).astype(np.uint8)
    H, W = mask.shape
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    if num_labels <= 1:
        return np.zeros_like(mask)

    max_y_reached = max(
        stats[i, cv2.CC_STAT_TOP] + stats[i, cv2.CC_STAT_HEIGHT] - 1
        for i in range(1, num_labels)
    )

    bottom_connected_mask = np.zeros_like(mask)
    for i in range(1, num_labels):
        comp_bottom = stats[i, cv2.CC_STAT_TOP] + stats[i, cv2.CC_STAT_HEIGHT] - 1
        if comp_bottom >= max(H - 1 - bottom_margin, max_y_reached - bottom_margin):
            bottom_connected_mask[labels == i] = 1

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, close_kernel_size)
    closed = cv2.morphologyEx(bottom_connected_mask, cv2.MORPH_CLOSE, kernel)

    final_mask = np.zeros_like(closed)
    for col in range(W):
        wet_rows = np.where(closed[:, col] == 1)[0]
        if len(wet_rows) == 0:
            continue
        bottom_pixel = wet_rows[-1]
        waterline = bottom_pixel
        while waterline >= 0 and closed[waterline, col] == 1:
            waterline -= 1
        waterline += 1
        final_mask[waterline : bottom_pixel + 1, col] = 1

    return final_mask


def compute_war(mask: np.ndarray, roi_mask: Optional[np.ndarray] = None) -> float:
    """Calculates Wetted Area Ratio: |M_wet and M_ROI| / |M_ROI|."""
    if roi_mask is not None:
        roi_count = np.sum(roi_mask > 0)
        return float(np.sum((mask > 0) & (roi_mask > 0)) / max(roi_count, 1))
    return float(np.sum(mask > 0) / max(mask.size, 1))


def compute_war_metrics(pred_wars: np.ndarray, gt_wars: np.ndarray) -> Dict[str, float]:
    """Computes Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and genuine sklearn R^2 for WAR."""
    mae = float(np.mean(np.abs(pred_wars - gt_wars)))
    rmse = float(np.sqrt(np.mean((pred_wars - gt_wars) ** 2)))
    # Genuine scikit-learn r2_score as required
    r2 = float(r2_score(gt_wars, pred_wars))
    return {
        "war_mae": mae,
        "war_rmse": rmse,
        "war_r2": r2,
    }

criterion = BCEDiceLoss(bce_weight=0.5, dice_weight=0.5)
optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=3)
print("Loss, AdamW optimizer, and metrics initialized.")
```

## Section 8: Fast Smoke Test
Executes a 1-batch forward/backward pass on real data to verify:
- Data loading and device placement.
- Forward pass through SegFormer-B1.
- BCE + Dice Loss calculation.
- Backpropagation gradient stability and optimizer step.
- Metric calculation without NaN/Inf values.

```python
def run_smoke_test(model: nn.Module, loader: DataLoader, criterion: nn.Module) -> None:
    """Executes a 1-step forward/backward sanity check on SegFormer-B1."""
    model.train()
    images, masks, vids, stems = next(iter(loader))
    images, masks = images.to(DEVICE), masks.to(DEVICE)
    logits = model(images)
    loss = criterion(logits, masks)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()

    with torch.no_grad():
        probs = torch.sigmoid(logits)
        preds = (probs > PRED_THRESHOLD).float()
        metrics = compute_batch_metrics(preds, masks)

    assert not torch.isnan(loss), "Loss is NaN!"
    assert not torch.isinf(loss), "Loss is Inf!"
    print(f"SegFormer-B1 smoke test passed successfully!")
    print(f"  Input batch shape : {images.shape}")
    print(f"  Target mask shape : {masks.shape}")
    print(f"  Logits shape      : {logits.shape}")
    print(f"  Batch loss        : {loss.item():.4f}")
    print(f"  Batch IoU         : {metrics['iou']:.4f}")
    print(f"  Batch Dice        : {metrics['dice']:.4f}")

run_smoke_test(model, train_loader, criterion)
```

- Early stopping with patience = 6.
- Saves epoch metrics history to `evaluation/segformer_b1/metrics_train_val.csv`.

```python
def train_one_epoch(
    model: nn.Module,
    loader: DataLoader,
    optimizer: optim.Optimizer,
    criterion: nn.Module,
) -> Tuple[float, Dict[str, float]]:
    """Trains SegFormer-B1 for one complete epoch."""
    model.train()
    total_loss = 0.0
    accumulated_metrics = {"iou": 0.0, "dice": 0.0, "precision": 0.0, "recall": 0.0, "accuracy": 0.0}

    for images, masks, _, _ in loader:
        images, masks = images.to(DEVICE), masks.to(DEVICE)
        optimizer.zero_grad()
        logits = model(images)
        loss = criterion(logits, masks)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        with torch.no_grad():
            probs = torch.sigmoid(logits)
            preds = (probs > PRED_THRESHOLD).float()
            b_metrics = compute_batch_metrics(preds, masks)
            for k in accumulated_metrics:
                accumulated_metrics[k] += b_metrics[k]

    num_batches = len(loader)
    avg_loss = total_loss / max(num_batches, 1)
    avg_metrics = {k: v / max(num_batches, 1) for k, v in accumulated_metrics.items()}
    return avg_loss, avg_metrics


@torch.no_grad()
def evaluate_one_epoch(
    model: nn.Module,
    loader: DataLoader,
    criterion: nn.Module,
) -> Tuple[float, Dict[str, float]]:
    """Evaluates SegFormer-B1 across the validation split."""
    model.eval()
    total_loss = 0.0
    accumulated_metrics = {"iou": 0.0, "dice": 0.0, "precision": 0.0, "recall": 0.0, "accuracy": 0.0}

    for images, masks, _, _ in loader:
        images, masks = images.to(DEVICE), masks.to(DEVICE)
        logits = model(images)
        loss = criterion(logits, masks)

        total_loss += loss.item()
        probs = torch.sigmoid(logits)
        preds = (probs > PRED_THRESHOLD).float()
        b_metrics = compute_batch_metrics(preds, masks)
        for k in accumulated_metrics:
            accumulated_metrics[k] += b_metrics[k]

    num_batches = len(loader)
    avg_loss = total_loss / max(num_batches, 1)
    avg_metrics = {k: v / max(num_batches, 1) for k, v in accumulated_metrics.items()}
    return avg_loss, avg_metrics


# Execute Training Loop
best_val_iou = -1.0
best_model_path = MODEL_SAVE_DIR / "segformer_best.pth"
checkpoint_best_path = CHECKPOINT_DIR / "best_model.pth"
last_model_path = CHECKPOINT_DIR / "last_model.pth"
patience_counter = 0
history = []

print(f"Starting SegFormer-B1 (MiT-B1) training for up to {EPOCHS} epochs on {DEVICE}...")

for epoch in range(1, EPOCHS + 1):
    t0 = time.time()
    train_loss, train_m = train_one_epoch(model, train_loader, optimizer, criterion)
    val_loss, val_m = evaluate_one_epoch(model, val_loader, criterion)
    scheduler.step(val_m["iou"])
    epoch_time = time.time() - t0

    is_best = val_m["iou"] > best_val_iou
    if is_best:
        best_val_iou = val_m["iou"]
        patience_counter = 0
        checkpoint_data = {
            "epoch": epoch,
            "architecture": "SegFormer-B1-MiT-B1",
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_iou": best_val_iou,
            "val_metrics": val_m,
        }
        torch.save(checkpoint_data, best_model_path)
        torch.save(checkpoint_data, checkpoint_best_path)
        save_msg = f" -> Saved best model ({best_model_path.name})"
    else:
        patience_counter += 1
        save_msg = ""

    # Always save last checkpoint
    torch.save({
        "epoch": epoch,
        "architecture": "SegFormer-B1-MiT-B1",
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "last_val_iou": val_m["iou"],
    }, last_model_path)

    log_entry = {
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss,
        "train_iou": train_m["iou"],
        "val_iou": val_m["iou"],
        "train_dice": train_m["dice"],
        "val_dice": val_m["dice"],
        "train_precision": train_m["precision"],
        "val_precision": val_m["precision"],
        "train_recall": train_m["recall"],
        "val_recall": val_m["recall"],
        "time_s": epoch_time,
    }
    history.append(log_entry)

    print(
        f"Epoch {epoch:02d}/{EPOCHS:02d} | "
        f"Train Loss: {train_loss:.4f}  Val Loss: {val_loss:.4f} | "
        f"Train IoU: {train_m['iou']:.4f}  Val IoU: {val_m['iou']:.4f} | "
        f"Train Dice: {train_m['dice']:.4f}  Val Dice: {val_m['dice']:.4f} ({epoch_time:.1f}s){save_msg}"
    )

    if patience_counter >= EARLY_STOP_PATIENCE:
        print(f"\nEarly stopping triggered after {epoch} epochs (patience={EARLY_STOP_PATIENCE}).")
        break

# Save Training Metrics History
train_history_df = pd.DataFrame(history)
train_history_df.to_csv(MODEL_SAVE_DIR / "metrics_train_val.csv", index=False)
print(f"\nBest Validation IoU: {best_val_iou:.4f}")
print(f"Saved epoch metrics log to: {MODEL_SAVE_DIR / 'metrics_train_val.csv'}")
```

## Section 12: Qualitative Visualizations & Physical WAR Trajectory Curves
Generates standardized research-grade qualitative and quantitative evaluation figures:
1. **Capillary Absorption WAR Trajectory Curves**: Front rise ($WAR$ vs. frame number) across each untouched test video comparing Ground Truth, Raw SegFormer-B1, and Physics-Enforced predictions.
2. **Standardized Five-Panel Qualitative Visualizations (Identical to U-Net++ Benchmark)**:
   - **Panel 1 — Input Frame**: Natural RGB image with specimen ID and filename title (`{video} - {stem}\nInput Frame`).
   - **Panel 2 — Ground Truth**: Binary consensus mask (white wet, black dry) with ground-truth $WAR$ value (`Ground Truth\nWAR: 0.xxx`).
   - **Panel 3 — Raw SegFormer-B1**: Binary prediction from bilinear-interpolated logits thresholded at $0.5$ (`Raw SegFormer-B1\nWAR: 0.xxx`).
   - **Panel 4 — Physics Enforced**: Waterline contiguity and column-monotonicity constrained mask (`Physics Enforced\nWAR: 0.xxx`).
   - **Panel 5 — Error Map**: Pixelwise classification map (`Green: True Positive`, `Red: False Positive`, `Blue: False Negative`, `Black: True Negative`). Title: `Error Map\n(G:TP, R:FP, B:FN)`.
3. **Reusable Inference & Visualization Function**: `visualize_segformer_prediction(...)` supporting arbitrary inputs, auto-computing $WAR$ values, and exporting 300 DPI figures.
4. **Automated Batch Test Set Export**: Saves 5-panel figures into `outputs/segformer_visualizations/`, exports `segformer_war_comparison.csv`, generates a summary contact sheet, and exports PDF sheets.

```python
# ==============================================================================
# 1. Plot 1: Capillary Absorption WAR Trajectory Curves
# ==============================================================================
test_df = pd.DataFrame(test_results)

fig, axes = plt.subplots(1, len(test_videos), figsize=(6 * len(test_videos), 4.5), sharey=True)
if len(test_videos) == 1:
    axes = [axes]

for ax, vid in zip(axes, test_videos):
    vid_data = test_df[test_df["video"] == vid].copy()
    vid_data["frame_num"] = vid_data["stem"].apply(lambda x: int(x.split("-")[-1]) if "-" in x else 0)
    vid_data = vid_data.sort_values("frame_num")

    ax.plot(vid_data["frame_num"], vid_data["war_gt"], label="Ground Truth (Consensus)", color="black", lw=2)
    ax.plot(vid_data["frame_num"], vid_data["war_pred_raw"], label="SegFormer-B1 (Raw)", color="#1f77b4", linestyle="--", lw=1.8)
    ax.plot(vid_data["frame_num"], vid_data["war_pred_phys"], label="SegFormer-B1 (Physics-Enforced)", color="#2ca02c", linestyle="-.", lw=1.8)

    ax.set_title(f"{vid.title()} WAR Trajectory (SegFormer-B1)", fontsize=12, fontweight="bold")
    ax.set_xlabel("Video Frame Index", fontsize=10)
    ax.set_ylabel("Wetted Area Ratio (WAR)", fontsize=10)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper left", fontsize=8)

plt.tight_layout()
trajectory_plot_path = MODEL_SAVE_DIR / "segformer_war_trajectories.png"
plt.savefig(trajectory_plot_path, dpi=300)
plt.show()
plt.close(fig)
print(f"Saved WAR trajectory curves to: {trajectory_plot_path}")

# ==============================================================================
# 2. Reusable Five-Panel Qualitative Visualization Function
# ==============================================================================
def denormalize_image(image: Union[np.ndarray, torch.Tensor, Image.Image]) -> np.ndarray:
    """Prepares RGB image array in [0.0, 1.0] for matplotlib display."""
    if isinstance(image, Image.Image):
        return np.array(image.convert("RGB"), dtype=np.float32) / 255.0
    if isinstance(image, torch.Tensor):
        img_np = image.detach().cpu().float()
        if img_np.ndim == 3 and img_np.shape[0] == 3:
            img_np = img_np.permute(1, 2, 0).numpy()
            if img_np.min() < -0.1:
                img_np = img_np * np.array(IMAGENET_STD) + np.array(IMAGENET_MEAN)
            return np.clip(img_np, 0.0, 1.0)
        elif img_np.ndim == 2:
            return np.clip(img_np.numpy(), 0.0, 1.0)
    img_np = np.array(image, dtype=np.float32)
    if img_np.max() > 1.0:
        img_np = img_np / 255.0
    return np.clip(img_np, 0.0, 1.0)


def visualize_segformer_prediction(
    image: Union[np.ndarray, torch.Tensor, Image.Image],
    ground_truth: Union[np.ndarray, torch.Tensor],
    raw_prediction: Union[np.ndarray, torch.Tensor],
    physics_prediction: Optional[Union[np.ndarray, torch.Tensor]],
    sample_name: str,
    save_path: Optional[Union[str, Path]] = None,
    show_fig: bool = False,
    figsize: Tuple[float, float] = (18.0, 3.5),
    dpi: int = 300,
) -> Dict[str, Any]:
    """Generates the standardized 5-panel qualitative segmentation visualization (identical to U-Net++ layout).
    
    Panel 1: Input Frame (Preserves natural RGB colors, specimen ID & filename title)
    Panel 2: Ground Truth (Binary mask, white wet / black non-wet, WAR title)
    Panel 3: Raw SegFormer-B1 (Binary prediction, WAR title)
    Panel 4: Physics Enforced (Waterline-constrained prediction, WAR title)
    Panel 5: Error Map (Green: True Positive, Red: False Positive, Blue: False Negative, Black: True Negative)
    """
    # 1. Format image array for natural RGB display
    img_disp = denormalize_image(image)

    # 2. Format ground truth and predictions to 2D binary uint8 arrays
    if isinstance(ground_truth, torch.Tensor):
        gt = (ground_truth.detach().cpu().squeeze().numpy() > 0.5).astype(np.uint8)
    else:
        gt = (np.squeeze(ground_truth) > 0.5).astype(np.uint8)

    if isinstance(raw_prediction, torch.Tensor):
        p_raw = (raw_prediction.detach().cpu().squeeze().numpy() > 0.5).astype(np.uint8)
    else:
        p_raw = (np.squeeze(raw_prediction) > 0.5).astype(np.uint8)

    if physics_prediction is not None:
        if isinstance(physics_prediction, torch.Tensor):
            p_phys = (physics_prediction.detach().cpu().squeeze().numpy() > 0.5).astype(np.uint8)
        else:
            p_phys = (np.squeeze(physics_prediction) > 0.5).astype(np.uint8)
    else:
        p_phys = None

    # 3. Calculate full-frame WAR values independently
    war_gt = compute_war(gt)
    war_raw = compute_war(p_raw)
    war_phys = compute_war(p_phys) if p_phys is not None else float("nan")

    error_raw = abs(war_raw - war_gt)
    error_phys = abs(war_phys - war_gt) if p_phys is not None else float("nan")

    # 4. Generate Error Map (G: TP, R: FP, B: FN, Black: TN) from final physics-enforced prediction
    eval_mask = p_phys if p_phys is not None else p_raw
    error_map = np.zeros((*gt.shape, 3), dtype=np.float32)
    error_map[(eval_mask == 1) & (gt == 1)] = [0.0, 1.0, 0.0]  # Green: TP
    error_map[(eval_mask == 1) & (gt == 0)] = [1.0, 0.0, 0.0]  # Red: FP
    error_map[(eval_mask == 0) & (gt == 1)] = [0.0, 0.0, 1.0]  # Blue: FN
    # TN remains [0, 0, 0] (Black)

    # 5. Create 5-panel horizontal figure
    fig, axes = plt.subplots(1, 5, figsize=figsize, facecolor="white")

    # Panel 1: Input Frame
    axes[0].imshow(img_disp)
    axes[0].set_title(f"{sample_name}\nInput Frame", fontsize=10)
    axes[0].axis("off")

    # Panel 2: Ground Truth
    axes[1].imshow(gt, cmap="gray", vmin=0, vmax=1)
    axes[1].set_title(f"Ground Truth\nWAR: {war_gt:.3f}", fontsize=10)
    axes[1].axis("off")

    # Panel 3: Raw SegFormer-B1
    axes[2].imshow(p_raw, cmap="gray", vmin=0, vmax=1)
    axes[2].set_title(f"Raw SegFormer-B1\nWAR: {war_raw:.3f}", fontsize=10)
    axes[2].axis("off")

    # Panel 4: Physics Enforced
    if p_phys is not None:
        axes[3].imshow(p_phys, cmap="gray", vmin=0, vmax=1)
        axes[3].set_title(f"Physics Enforced\nWAR: {war_phys:.3f}", fontsize=10)
    else:
        axes[3].text(0.5, 0.5, "Physics Enforced\n(Unavailable)", ha="center", va="center", color="gray")
        axes[3].set_title("Physics Enforced\nWAR: N/A", fontsize=10)
    axes[3].axis("off")

    # Panel 5: Error Map
    axes[4].imshow(error_map)
    axes[4].set_title("Error Map\n(G:TP, R:FP, B:FN)", fontsize=10)
    axes[4].axis("off")

    plt.tight_layout()

    if save_path is not None:
        save_p = Path(save_path)
        save_p.parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(save_p, dpi=dpi, facecolor="white", bbox_inches="tight")

    if show_fig:
        plt.show()
    plt.close(fig)

    return {
        "sample_name": sample_name,
        "war_gt": war_gt,
        "war_pred_raw": war_raw,
        "war_pred_phys": war_phys,
        "error_raw": error_raw,
        "error_phys": error_phys,
    }

# ==============================================================================
# 3. Plot 2: Sample Qualitative Visualizations (Exact 5-Panel U-Net++ Layout)
# ==============================================================================
sample_indices = [0, len(test_loader.dataset) // 2, len(test_loader.dataset) - 1]
fig, axes = plt.subplots(len(sample_indices), 5, figsize=(18, 3.5 * len(sample_indices)), facecolor="white")

for row, idx in enumerate(sample_indices):
    img_t, mask_t, vid_name, stem = test_loader.dataset[idx]
    with torch.no_grad():
        logit = model(img_t.unsqueeze(0).to(DEVICE))
        gt = mask_t[0].numpy().astype(np.uint8)
        # Ensure bilinear interpolation to ground truth mask dimensions
        if logit.shape[-2:] != gt.shape:
            logit = torch.nn.functional.interpolate(logit, size=gt.shape, mode="bilinear", align_corners=False)
        prob = torch.sigmoid(logit)[0, 0].cpu().numpy()
        p_raw = (prob > PRED_THRESHOLD).astype(np.uint8)
        p_phys = enforce_waterline_physics(p_raw, bottom_margin=BOTTOM_MARGIN)

    # Denormalize image for display
    img_np = img_t.permute(1, 2, 0).numpy()
    img_disp = np.clip(img_np * IMAGENET_STD + IMAGENET_MEAN, 0, 1)

    # Error Map: Green = True Positive, Red = False Positive, Blue = False Negative
    error_map = np.zeros((*gt.shape, 3), dtype=np.float32)
    error_map[(p_phys == 1) & (gt == 1)] = [0, 1, 0]  # TP (Green)
    error_map[(p_phys == 1) & (gt == 0)] = [1, 0, 0]  # FP (Red)
    error_map[(p_phys == 0) & (gt == 1)] = [0, 0, 1]  # FN (Blue)

    axes[row, 0].imshow(img_disp); axes[row, 0].set_title(f"{vid_name} - {stem}\nInput Frame", fontsize=10)
    axes[row, 1].imshow(gt, cmap="gray"); axes[row, 1].set_title(f"Ground Truth\nWAR: {compute_war(gt):.3f}", fontsize=10)
    axes[row, 2].imshow(p_raw, cmap="gray"); axes[row, 2].set_title(f"Raw SegFormer-B1\nWAR: {compute_war(p_raw):.3f}", fontsize=10)
    axes[row, 3].imshow(p_phys, cmap="gray"); axes[row, 3].set_title(f"Physics Enforced\nWAR: {compute_war(p_phys):.3f}", fontsize=10)
    axes[row, 4].imshow(error_map); axes[row, 4].set_title("Error Map\n(G:TP, R:FP, B:FN)", fontsize=10)

    for c in range(5):
        axes[row, c].axis("off")

plt.tight_layout()
visual_comparison_path = MODEL_SAVE_DIR / "segformer_visual_predictions.png"
plt.savefig(visual_comparison_path, dpi=300, facecolor="white")
plt.show()
plt.close(fig)
print(f"Saved visual prediction panels to: {visual_comparison_path}")

# ==============================================================================
# 4. Automated Batch Test Evaluation & Output Generation
# ==============================================================================
from matplotlib.backends.backend_pdf import PdfPages

viz_output_dir = Path("outputs/segformer_visualizations")
viz_output_dir.mkdir(parents=True, exist_ok=True)

print(f"Generating standardized 5-panel visualizations for test set into: {viz_output_dir.resolve()}")
batch_records = []
pdf_export_path = viz_output_dir / "segformer_test_visualizations.pdf"
rep_indices = [0, 15, 30, 45, 63, 64, 90, 120, 150, 177, 178, 190, 205, 217]

with PdfPages(pdf_export_path) as pdf:
    for i, (img_t, mask_t, vid_name, stem) in enumerate(test_loader.dataset):
        with torch.no_grad():
            logit = model(img_t.unsqueeze(0).to(DEVICE))
            gt = mask_t[0].numpy().astype(np.uint8)
            if logit.shape[-2:] != gt.shape:
                logit = torch.nn.functional.interpolate(logit, size=gt.shape, mode="bilinear", align_corners=False)
            prob = torch.sigmoid(logit)[0, 0].cpu().numpy()
            p_raw = (prob > PRED_THRESHOLD).astype(np.uint8)
            p_phys = enforce_waterline_physics(p_raw, bottom_margin=BOTTOM_MARGIN)

        vid_clean = vid_name.replace(" ", "_")
        frame_part = stem.replace("ezgif-", "").replace("-", "_")
        sample_filename = f"{vid_clean}_{frame_part}_segformer.png"
        sample_save_path = viz_output_dir / sample_filename
        sample_display_name = f"{vid_name} - {stem}"

        metrics_res = visualize_segformer_prediction(
            image=img_t,
            ground_truth=gt,
            raw_prediction=p_raw,
            physics_prediction=p_phys,
            sample_name=sample_display_name,
            save_path=sample_save_path,
            show_fig=False,
            dpi=300,
        )
        metrics_res["video"] = vid_name
        metrics_res["stem"] = stem
        metrics_res["filename"] = sample_filename
        batch_records.append(metrics_res)

        if i in rep_indices:
            fig_pdf, ax_pdf = plt.subplots(1, 5, figsize=(18.0, 3.5), facecolor="white")
            img_disp = denormalize_image(img_t)
            err_m = np.zeros((*gt.shape, 3), dtype=np.float32)
            err_m[(p_phys == 1) & (gt == 1)] = [0, 1, 0]
            err_m[(p_phys == 1) & (gt == 0)] = [1, 0, 0]
            err_m[(p_phys == 0) & (gt == 1)] = [0, 0, 1]
            ax_pdf[0].imshow(img_disp); ax_pdf[0].set_title(f"{sample_display_name}\nInput Frame", fontsize=10); ax_pdf[0].axis("off")
            ax_pdf[1].imshow(gt, cmap="gray"); ax_pdf[1].set_title(f"Ground Truth\nWAR: {metrics_res['war_gt']:.3f}", fontsize=10); ax_pdf[1].axis("off")
            ax_pdf[2].imshow(p_raw, cmap="gray"); ax_pdf[2].set_title(f"Raw SegFormer-B1\nWAR: {metrics_res['war_pred_raw']:.3f}", fontsize=10); ax_pdf[2].axis("off")
            ax_pdf[3].imshow(p_phys, cmap="gray"); ax_pdf[3].set_title(f"Physics Enforced\nWAR: {metrics_res['war_pred_phys']:.3f}", fontsize=10); ax_pdf[3].axis("off")
            ax_pdf[4].imshow(err_m); ax_pdf[4].set_title("Error Map\n(G:TP, R:FP, B:FN)", fontsize=10); ax_pdf[4].axis("off")
            plt.tight_layout()
            pdf.savefig(fig_pdf, dpi=300, bbox_inches="tight")
            plt.close(fig_pdf)

# Export WAR Comparison CSV
war_comparison_df = pd.DataFrame(batch_records)[
    ["video", "stem", "filename", "sample_name", "war_gt", "war_pred_raw", "war_pred_phys", "error_raw", "error_phys"]
]
war_csv_path = viz_output_dir / "segformer_war_comparison.csv"
war_comparison_df.to_csv(war_csv_path, index=False)
print(f"Saved WAR comparison CSV to: {war_csv_path} ({len(war_comparison_df)} samples)")

# Export Summary Contact Sheet
contact_indices = [0, 30, 64, 120, 177, 198]
fig_c, axes_c = plt.subplots(len(contact_indices), 5, figsize=(18, 3.2 * len(contact_indices)), facecolor="white")
for r, c_idx in enumerate(contact_indices):
    img_t, mask_t, vid_name, stem = test_loader.dataset[c_idx]
    with torch.no_grad():
        logit = model(img_t.unsqueeze(0).to(DEVICE))
        gt = mask_t[0].numpy().astype(np.uint8)
        if logit.shape[-2:] != gt.shape:
            logit = torch.nn.functional.interpolate(logit, size=gt.shape, mode="bilinear", align_corners=False)
        prob = torch.sigmoid(logit)[0, 0].cpu().numpy()
        p_raw = (prob > PRED_THRESHOLD).astype(np.uint8)
        p_phys = enforce_waterline_physics(p_raw, bottom_margin=BOTTOM_MARGIN)

    img_disp = denormalize_image(img_t)
    w_gt = compute_war(gt)
    w_raw = compute_war(p_raw)
    w_phys = compute_war(p_phys)
    err_m = np.zeros((*gt.shape, 3), dtype=np.float32)
    err_m[(p_phys == 1) & (gt == 1)] = [0, 1, 0]
    err_m[(p_phys == 1) & (gt == 0)] = [1, 0, 0]
    err_m[(p_phys == 0) & (gt == 1)] = [0, 0, 1]

    axes_c[r, 0].imshow(img_disp); axes_c[r, 0].set_title(f"{vid_name} - {stem}\nInput Frame", fontsize=9.5); axes_c[r, 0].axis("off")
    axes_c[r, 1].imshow(gt, cmap="gray"); axes_c[r, 1].set_title(f"Ground Truth\nWAR: {w_gt:.3f}", fontsize=9.5); axes_c[r, 1].axis("off")
    axes_c[r, 2].imshow(p_raw, cmap="gray"); axes_c[r, 2].set_title(f"Raw SegFormer-B1\nWAR: {w_raw:.3f}", fontsize=9.5); axes_c[r, 2].axis("off")
    axes_c[r, 3].imshow(p_phys, cmap="gray"); axes_c[r, 3].set_title(f"Physics Enforced\nWAR: {w_phys:.3f}", fontsize=9.5); axes_c[r, 3].axis("off")
    axes_c[r, 4].imshow(err_m); axes_c[r, 4].set_title("Error Map\n(G:TP, R:FP, B:FN)", fontsize=9.5); axes_c[r, 4].axis("off")

plt.tight_layout()
contact_sheet_path = viz_output_dir / "segformer_contact_sheet.png"
plt.savefig(contact_sheet_path, dpi=300, facecolor="white", bbox_inches="tight")
plt.close(fig_c)
print(f"Saved summary contact sheet to: {contact_sheet_path}")
print(f"Saved representative PDF export to: {pdf_export_path}")
```
